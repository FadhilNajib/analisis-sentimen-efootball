```python
from google_play_scraper import Sort, reviews

# Variable menyiimpan ID aplikasi eFootball di Google Play Store
app_id = 'jp.konami.pesam'

print("Memulai proses scraping ulasan eFootball....")

# Proses pengambilan data
result, _ = reviews(
    app_id,
    lang='id',         # Bahasa Indonesia
    country='id',      # Regional Indonesia
    sort=Sort.NEWEST,  # Mengambil ulasan terbaru
    count=12000        # Target jumlah data
)

```

```python
# Mengubah list ulasan menjadi DataFrame
df = pd.DataFrame(result)

# Memfilter kolom: hanya ulasan (content) dan rating (score)
df = df[['content', 'score']]

# Mengganti nama kolom agar lebih informatif
df.columns = ['Review', 'Rating']

# Menghapus ulasan yang duplikat
df = df.drop_duplicates()

# Menyimpan hasil ke file CSV tanpa index (angka urutan di sebelah kiri)
df.to_csv('dataset_efootball_raw.csv', index=False, encoding='utf-8')

print(f"Scraping selesai! {len(df)} data berhasil disimpan di 'dataset_efootball_raw.csv'.")
```
